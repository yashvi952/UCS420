```python

import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(10, 6))

plt.plot(x, y1, color='blue', linestyle='-',
         label='y = M*x^2')

plt.plot(x, y2, color='red', linestyle='--',
         label='y = M*sin(x)')

plt.title("Plot of Mathematical Functions")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()
plt.grid(True)

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Subject': ['Maths', 'Physics', 'Chemistry',
                'English', 'Computer Science'],
    'Score': [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data)

print(df)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=df,
    x='Subject',
    y='Score',
    hue='Subject',
    palette='viridis',
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)

plt.title("Scores of Five Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.ylim(0, 110)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1024170157)

data = np.random.randn(50)
x = np.arange(1, 51)

cumulative_sum = np.cumsum(data)
noise = np.random.randn(50) * 0.3

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# Plot 1: Cumulative sum
axes[0, 0].plot(x, cumulative_sum, color='blue')
axes[0, 0].set_title("Cumulative Sum")
axes[0, 0].set_xlabel("Data Index")
axes[0, 0].set_ylabel("Cumulative Sum")
axes[0, 0].grid(True)

# Plot 2: Scatter plot with noise
axes[0, 1].scatter(x, data + noise, color='red')
axes[0, 1].set_title("Scatter Plot with Random Noise")
axes[0, 1].set_xlabel("Data Index")
axes[0, 1].set_ylabel("Value + Noise")
axes[0, 1].grid(True)

# Plot 3: Original random values
axes[1, 0].plot(x, data, color='green')
axes[1, 0].set_title("Original Random Data")
axes[1, 0].set_xlabel("Data Index")
axes[1, 0].set_ylabel("Random Value")
axes[1, 0].grid(True)

# Plot 4: Distribution of random values
axes[1, 1].hist(data, bins=10, color='orange',
                edgecolor='black')
axes[1, 1].set_title("Distribution of Random Data")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=df,
    x='month_number',
    y='total_profit',
    marker='o',
    color='blue'
)

plt.title("Monthly Total Profit")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.xticks(df['month_number'])
plt.grid(True)

plt.show()
```

```python

products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

plt.figure(figsize=(12, 7))

for product in products:
    sns.lineplot(
        data=df,
        x='month_number',
        y=product,
        marker='o',
        label=product
    )

plt.title("Monthly Sales of All Products")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")
plt.xticks(df['month_number'])
plt.legend(title="Products")
plt.grid(True)

plt.show()
```

```python

features = df.columns.drop('month_number')

fig, axes = plt.subplots(4, 2, figsize=(15, 18))

axes = axes.flatten()

for i, feature in enumerate(features):

    sns.barplot(
        data=df,
        x='month_number',
        y=feature,
        ax=axes[i],
        color='steelblue'
    )

    axes[i].set_title(f"Monthly {feature}")
    axes[i].set_xlabel("Month")
    axes[i].set_ylabel(feature)
    axes[i].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
y=np.array([1,1,1,2,3,4,2,4,3,3])
```

```python
values,counts=np.unique(y,return_counts=True)
most_frequent=values[np.argmax(counts)]
indices=np.where(y==most_frequent)[0]
print("Most frequent value:",most_frequent)
print("Indices:",indices)
```

```python
values,counts=np.unique(y,return_counts=True)
max_count=np.max(counts)
most_frequent=values[counts==max_count]
print("Most frequent values:",most_frequent)
for value in most_frequent:
  print(value,":",np.where(y==value)[0])
```
