```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most Frequent Value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)
most_frequent = values[counts == max_count]

print("Most Frequent Values:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    indices = np.where(y == value)[0]
    print(f"Indices of {value}:", indices)
```

Question: 4

```python
yashvi = np.linspace(10, 100, 25)

print("Original Array:", yashvi)

print("Dimensions:", yashvi.ndim)
print("Shape:", yashvi.shape)
print("Total Elements:", yashvi.size)
print("Data Type:", yashvi.dtype)
print("Total Bytes:", yashvi.nbytes)

# Transpose using reshape()
transposed = yashvi.reshape(1, 25).T

print("Transpose using reshape():")
print(transposed)

# Transpose using T attribute
print("Transpose using T:")
print(yashvi.T)
```

```python
import numpy as np

ucs420_yashvi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_yashvi)

# Statistical operations
print("Mean:", np.mean(ucs420_yashvi))
print("Median:", np.median(ucs420_yashvi))
print("Maximum:", np.max(ucs420_yashvi))
print("Minimum:", np.min(ucs420_yashvi))
print("Unique Elements:", np.unique(ucs420_yashvi))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_yashvi = ucs420_yashvi.reshape(4, 3)

print("\nReshaped Array (4x3):")
print(reshaped_ucs420_yashvi)

# Resize to 2 rows and 3 columns
resized_ucs420_yashvi = np.resize(ucs420_yashvi, (2, 3))

print("\nResized Array (2x3):")
print(resized_ucs420_yashvi)
```
